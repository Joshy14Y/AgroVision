# 01: Metadata Creation & Data Ingestion

## Project: AgroVision
**Purpose:** Scan the raw file directory to index all available images, extract class labels from folder structures, and generate the initial metadata registry.

### Key Objectives:
1.  **File Indexing:** Recursively scan the raw data directory for valid image formats (`.jpg`, `.jpeg`, `.png`).
2.  **Label Extraction:** Derive ground-truth labels automatically from the parent directory names.
3.  **Registry Generation:** Compile a master CSV (`dataset.csv`) to serve as the single source of truth for downstream EDA and training tasks.

```python
from pathlib import Path

import pandas as pd
```

## 1. Path Configuration
Define the input (raw data) and output (processed) directory paths to ensure consistent file access across the project lifecycle.

```python
PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data"
RAW_DATA_PATH = DATA_PATH / "raw"
PROCESSED_DATA_PATH = DATA_PATH / "processed"
CSV_PATH = PROCESSED_DATA_PATH / "dataset.csv"
```

```python
image_data = []
valid_ext = {".jpg", ".jpeg", ".png"}
category_folders = [folder for folder in RAW_DATA_PATH.iterdir() if folder.is_dir()]
```

## 2. Directory Traversal & Indexing
The raw data directory is scanned recursively to identify valid image assets. Non-image files are discarded to ensure dataset purity, and labels are inferred from the folder hierarchy.

```python
for filepath in RAW_DATA_PATH.rglob('*'):
    if filepath.is_file() and filepath.suffix.lower() in valid_ext:
        image_data.append({
            'filepath': str(filepath.relative_to(RAW_DATA_PATH)),
            'label': filepath.parent.name
        })
print(f"Found {len(image_data)} valid images.")
```

```python
df = pd.DataFrame(image_data)
```

## 3. Metadata Serialization
The structured metadata is exported to a CSV file. This file acts as the uncleaned "Raw Record" that will be processed in the EDA stage.

```python
df.to_csv(CSV_PATH, index=False)
```

## 4. Summary & Next Steps
The raw data ingestion is complete. All image paths have been indexed and labeled based on the directory structure.

**Key Outcomes:**
1.  **Registry Created:** A raw metadata file (`dataset.csv`) has been generated containing the file paths and initial labels.
2.  **Classes Identified:** The directory structure was successfully parsed to extract the class labels.

**Next Steps:**
The generated CSV is now ready for **`02_eda.ipynb`**, where we will perform quality checks, cleaning, and data splitting.
