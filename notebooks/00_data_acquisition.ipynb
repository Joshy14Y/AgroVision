# 00: Data Acquisition

## Project: AgroVision
**Purpose:** Retrieve the raw image dataset from external sources (Kaggle) and populate the project's local directory structure.

### Key Objectives:
1.  **Authentication:** Initialize Kaggle API credentials.
2.  **Retrieval:** Fetch the latest version of the dataset to a local cache.
3.  **Persistence:** Transfer the data to the canonical `data/raw` directory for immutable storage.

```python
import shutil
from pathlib import Path

import kagglehub
from dotenv import load_dotenv
```

```python
load_dotenv()
```

## 1. Environment & Path Configuration
Define the dataset identifier and the local directory structure where raw assets will be stored.

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"
DATASET_NAME = "raghavrpotdar/fresh-and-stale-images-of-fruits-and-vegetables"
```

## 2. External Data Retrieval
The dataset is fetched using the Kaggle API (`kagglehub`). This downloads the artifacts to a temporary local cache managed by the library.

```python
path_cache = Path(kagglehub.dataset_download(DATASET_NAME))
```

## 3. Local Storage & Persistence
The cached data is copied to the project's `data/raw` directory. This ensures the project has a stable, self-contained copy of the inputs, independent of the cache state.

```python
shutil.copytree(path_cache, RAW_DATA_PATH, dirs_exist_ok=True)
```

## 4. Summary & Next Steps
The data acquisition phase is complete. The raw images are successfully stored locally.

**Key Outcomes:**
1.  **Data Downloaded:** The `fresh-and-stale-images-of-fruits-and-vegetables` dataset was retrieved.
2.  **Storage Populated:** The `data/raw` directory now contains the image files.
